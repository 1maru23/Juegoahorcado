**Question 1:** How many customers are in the file?
(as help, we have added some comments and starter code to help you structure the solution)

```python
from os.path import exists
import csv

datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

total_customers = 0

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    next(csvreader)  # Skip header row
    for row in csvreader:
        if len(row) < 5 or not any(row):
            continue
        total_customers += 1

print(f"Q1 - Total customers: {total_customers}")
```

**Question 2:** In how many different states do the customers live in?

```python
from os.path import exists
import csv

datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

states = {}

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    next(csvreader)  # Skip header row
    for row in csvreader:
        if len(row) < 5 or not any(row):
            continue
        state = row[4].strip()
        if state:
            states[state] = states.get(state, 0) + 1

print(f"Q2 - Number of different states: {len(states)}")
```

**Question 3** What is the state with most customers?

```python
from os.path import exists
import csv

datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

states = {}

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    next(csvreader)  # Skip header row
    for row in csvreader:
        if len(row) < 5 or not any(row):
            continue
        state = row[4].strip()
        if state:
            states[state] = states.get(state, 0) + 1

most_customers_state = max(states, key=states.get)
print(f"Q3 - State with most customers: {most_customers_state} ({states[most_customers_state]} customers)")
```

**Question 4** What is the state with the least customers?

```python
from os.path import exists
import csv

datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

states = {}

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    next(csvreader)  # Skip header row
    for row in csvreader:
        if len(row) < 5 or not any(row):
            continue
        state = row[4].strip()
        if state:
            states[state] = states.get(state, 0) + 1

# Find the minimum value
min_count = min(states.values())

# Get all states with that minimum
least_states = [state for state, count in states.items() if count == min_count]

print(f"Q4 - States with least customers ({min_count} customer each): {least_states}")
```

**Question 5:** What is the most common last name?

The second sample files contains orders placed by customers from the first file. Be careful, this file has many rows and you most likely should not print the contents of the whole file.

**Question #1:** How many unique orders are in the orders.csv file?

**Question #2:** What is the average number of items per order (rounded to two decimal places)?

**Question #3:** What is the highest number of items per order?

**Question #4:** What is the number of orders placed in October 2021?

**Question #5:** Which customer spent the most amount of money in 2021?

**Question #6:** Historically, what is the best month for sales?

Once you get your answers, remember to go back to the course and introduce them in the multiple choice quiz

```python
from os.path import exists
import csv

datafile = "./sample_data/orders.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

# Initialize variables
orders = {}
monthly_sales = {}

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    next(csvreader)  # Skip header row

    for row in csvreader:
        if len(row) < 6:
            continue

        customer_id = row[0].strip()
        order_id    = row[1].strip()
        date        = row[2].strip()
        product     = row[4].strip()
        price       = row[5].strip()

        if not order_id:
            continue

        # Build orders dictionary
        if order_id not in orders:
            orders[order_id] = {
                "customer_id": customer_id,
                "date": date,
                "total": 0.0,
                "items": []
            }

        # Add item to order
        if product:
            orders[order_id]["items"].append(product)

        # Add price to order total
        try:
            orders[order_id]["total"] += float(price)
        except ValueError:
            pass

        # Track monthly sales
        try:
            month_key = date[:7]
            if month_key and len(month_key) == 7:  # Must be format YYYY-MM
                monthly_sales[month_key] = monthly_sales.get(month_key, 0.0) + float(price)
        except ValueError:
            pass

# Q1: Total unique orders
total_orders = len(orders)
print(f"Q1 - Total unique orders: {total_orders}")

# Q2: Average number of items per order
avg_items = sum(len(o["items"]) for o in orders.values()) / total_orders
print(f"Q2 - Average items per order: {round(avg_items, 2)}")

# Q3: Highest number of items in a single order
max_items = max(len(o["items"]) for o in orders.values())
print(f"Q3 - Highest number of items per order: {max_items}")

# Q4: Orders placed in October 2021
oct_2021_orders = sum(1 for o in orders.values() if o["date"].startswith("2021-10"))
print(f"Q4 - Orders in October 2021: {oct_2021_orders}")

# Q5: Customer who spent the most in 2021
spending_2021 = {}
for o in orders.values():
    if o["date"].startswith("2021"):
        cid = o["customer_id"]
        spending_2021[cid] = spending_2021.get(cid, 0.0) + o["total"]

top_customer = max(spending_2021, key=spending_2021.get)
print(f"Q5 - Top spending customer in 2021: {top_customer} (${spending_2021[top_customer]:.2f})")

# Q6: Best month for sales historically
best_month = max(monthly_sales, key=monthly_sales.get)
print(f"Q6 - Best month for sales: {best_month} (${monthly_sales[best_month]:.2f})")
```
